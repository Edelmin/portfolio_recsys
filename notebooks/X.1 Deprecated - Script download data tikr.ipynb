# Reformateo de los datos

```python
def market_value(string):

    if 'US$MM' in string:
        return str(float(string[:-6])/1000)

    if 'US$B' in string:
        return string[:-5]




import os

dir_path = 'data/02_intermediate/DatosSectoresExtraido/'
dest_dir_path = 'data/02_intermediate/DatosSectores/'

for file in os.listdir(dir_path):

    if file.endswith('.csv'):

        df = pd.read_csv(dir_path+file)
        print(file)
        df = df[df['Valor de Mercado'].apply(lambda x: str(x).lower() != 'nan')]

        if (df['Valor de Mercado'].apply(lambda x: ('US$B' in str(x))  or ('US$MM' in str(x))).sum() - df.shape[0])== 0:
            print("Valor de mercado en dólares americanos.")
                        
            df['Valor de Mercado'] = df['Valor de Mercado'].apply(market_value)

            df.rename(columns={"Valor de Mercado": "Valor de mercado (US$B)"}, inplace=True)

            df = df[['Símbolo', 'Empresa', 'Industria','Valor de mercado (US$B)', 'Sector', 
                'Ubicación']]
            df.to_csv(dest_dir_path+file, index=False)

        else:
            print("El dataframe no mide el valor de mercado en dólares americanos.")
            break
        
        print("Se terminó la comprobación satisfactoriamente.")
```

```python
import pandas as pd
import os
from bs4 import BeautifulSoup
import pandas as pd

def load_html(path='path.txt'):

    with open(path, "r", encoding='utf-8') as file:
        html = file.read()
    return html

def load_dataframe_from_csv(file):
    
    return pd.read_csv(file)


def get_href_related(soup="html_parser"):

    # Almacenar filas
    data = []

    # Recorrer cada fila de la tabla
    for tr in soup.find_all('tr'):
        tds = tr.find_all('td')
        
        if len(tds) >= 3:
            # Segundo <td> (índice 1) tiene el símbolo con el primer <a>
            symbol_tag = tds[1].find('a')
            symbol = symbol_tag.get_text(strip=True) if symbol_tag else ''
            href = symbol_tag['href'] if symbol_tag and symbol_tag.has_attr('href') else ''
            
            # Tercer <td> (índice 2) tiene el nombre de la empresa
            name_tag = tds[2].find('a')
            name = name_tag.get_text(strip=True) if name_tag else ''
            
            # Guardar la fila si el href empieza con /stock
            if href.startswith('/stock'):
                data.append({
                    'Símbolo': symbol,
                    'Empresa': name,
                    'Enlace': href
                })
    return pd.DataFrame(data)



def get_dataframe_with_href_from_txt(html_path):

    html = load_html(html_path)
    soup = BeautifulSoup(html, 'html.parser')

    df = get_href_related(soup)
    return df


# small_caps_path = "data/02_intermediate/DatosSectores/SmallCaps/"
# large_caps_path = "data/02_intermediate/DatosSectores/LargeCaps/"

# small_caps_file_paths = [small_caps_path + file for file in os.listdir(small_caps_path)]
# large_caps_file_paths = [large_caps_path + file for file in os.listdir(large_caps_path)]

# small_caps_file_names = [file[:-4] for file in os.listdir(small_caps_path)]
# large_caps_file_names = [file[:-4] for file in os.listdir(large_caps_path)]

file_path = "data/02_intermediate/DatosSectoresExtraido/"

file_names = [file[:-4] for file in os.listdir(file_path)]
csv_path_list = [file_path + file for file in os.listdir(file_path)]



```

```python
pd.read_csv(csv_path_list[0])
```

```python
txt_dir_path = "data/01_raw/FicherosHtmlSectores/"

for csv_file, name_file in zip(csv_path_list, file_names):
    for txt_file in os.listdir(txt_dir_path):
        if name_file in txt_file:
            
            html_df = get_dataframe_with_href_from_txt(txt_dir_path + txt_file)
            html_df.set_index("Símbolo", inplace=True)
            html_df['Enlace'] = html_df['Enlace'].apply(lambda x: x.replace("about", "financials"))
            html_df['Enlace'] = html_df['Enlace'].apply(lambda x: x.replace("/stock", "https://app.tikr.com/stock"))

            break
    
    
    df = pd.read_csv(csv_file)
    df = df.join(html_df['Enlace'], on='Símbolo')
    
    df.to_csv("data/02_intermediate/DatosSectoresConHref/" + name_file + '.csv', index=False)
    
    
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from dotenv import load_dotenv
import time
import os
import pandas as pd

load_dotenv()
os.chdir("C:/Users/Edelmín/Desktop/Estudios/Master y cursos universitarios/Inteligencia Artificial/Asignatura y apuntes/TFM/TFM/tfm_env")
def session(headless=False):
        
    # Ruta al ejecutable de Brave
    brave_path = os.environ['PATH_BRAVE']  # Windows

    # Ruta a ChromeDriver (ajusta según tu sistema)
    chromedriver_path = os.environ['PATH_CHROMEDRIVER']

    options = Options()
    options.binary_location = brave_path

    options.add_argument(r"user-data-dir=C:/Users/Edelmín/AppData/Local/BraveSoftware/Brave-Browser/User Data")
    options.add_argument("profile-directory=Default")  # o "Profile 1", según corresponda


    # Configuración de descarga automática
    prefs = {
        "download.default_directory": os.environ['PATH_DOWNLOAD'],
        "download.prompt_for_download": False,
        "download.directory_upgrade": True,
        "safebrowsing.enabled": True
    }
    options.add_experimental_option("prefs", prefs)


    if headless:
        options.add_argument("--headless")  # Para correr sin ventana

    # Crea el servicio y el navegador
    service = Service(chromedriver_path)
    driver = webdriver.Chrome(service=service, options=options)

    return service, driver


def open_url(driver, url):
    driver.get(url)

def locate_window(driver):

    driver.set_window_position(-1920, 0)  # Ajusta 1920 si tu resolución cambia

    driver.maximize_window()


def login(driver):
    
    email = WebDriverWait(driver, 3).until(
    EC.visibility_of_element_located((By.ID, "input-13"))
    )
    
    email.send_keys(os.environ['TIKR_EMAIL'])

    contraseña = WebDriverWait(driver, 3).until(
    EC.visibility_of_element_located((By.ID, "input-16"))
    )
    
    contraseña.send_keys(os.environ['TIKR_PASSWORD'])


    login_boton = WebDriverWait(driver, 10).until(
        EC.element_to_be_clickable((By.CSS_SELECTOR, "button[data-v-0c6014fc]"))
    )
    login_boton.click()

    

    cerrar_boton = WebDriverWait(driver, 10).until(
        EC.element_to_be_clickable((By.XPATH, "//button[.//i[contains(@class, 'mdi-close')]]"))
    )
    cerrar_boton.click()


def load_href_list(dir):
    
    d = pd.DataFrame()
    ordered_list = []
    for file in os.listdir(dir):
        df = pd.read_csv(dir + file)
        ordered_list += [file for i in range(len(df))]
    d = pd.concat([d, df])
        
    return d.Enlace.tolist(), d['Símbolo'].tolist(), ordered_list

def open_dataset_source(driver):

    dataset = WebDriverWait(driver, 10).until(EC.element_to_be_clickable((
    By.XPATH,
    "//div[@role='button' and @class='v-input__slot']"
    )))
    dataset.click()
    

def select_morningstar_source(driver):

    morningstar = WebDriverWait(driver, 5).until(EC.element_to_be_clickable((
    By.XPATH,
    "//div[@role='option'][div/div[text()='Morningstar (Beta)']]"
    )))
    morningstar.click()
    

def open_export_selector(driver): 

    export_buttom = WebDriverWait(driver, 5).until(EC.element_to_be_clickable((
    By.XPATH,
    "//div[@class='export-button']"
    )))
    export_buttom.click()





    
service, driver = session()
wait = WebDriverWait(driver, 10)

open_url(driver, "https://app.tikr.com/")
# locate_window(driver)
# login(driver)


href_list, company_list, file_list = load_href_list(os.environ['PATH_HREF'])
with open("data/01_raw/Descargas_incorrectas.txt", "a", encoding='utf-8') as errors,\
open("data/01_raw/Descargas correctas.txt", "a", encoding='utf-8') as corrects:
    
    errors.write("Enlace, Símbolo, Fichero\n")
    corrects.write("Enlace, Símbolo, Fichero\n")

for company_ref, company_ticker, file_container in zip(href_list, company_list, file_list):

    try:



        driver.get(company_ref)

        open_dataset_source(driver)
        select_morningstar_source(driver)
        open_export_selector(driver)
        
        
        statement_all = wait.until(EC.element_to_be_clickable((
        By.XPATH,
        "//legend[div[contains(normalize-space(.), 'Statement')]]/following-sibling::div//label[div[contains(normalize-space(.), 'All')]]")))
        statement_all.click()
        
        years_all = wait.until(EC.element_to_be_clickable((
        By.XPATH,
        "//legend[div[normalize-space(text())='Years']]/following-sibling::div//label[div[normalize-space(text())='All']]"
        )))
        years_all.click()
        
        
        export_buttom = wait.until(EC.element_to_be_clickable((
        By.CSS_SELECTOR,
        "button.btn-excel-download"
        )))
        export_buttom.click()
        with open("data/01_raw/Descargas correctas.txt", "a", encoding='utf-8') as corrects:
            corrects.write(f"{company_ref},{company_ticker},{file_container}\n")
        
    except Exception as e:
        with open("data/01_raw/Descargas_incorrectas.txt", "a", encoding='utf-8') as errors:
            errors.write(f"{company_ref},{company_ticker},{file_container}\n")
        continue

print('Terminé la ejecución.')
```
