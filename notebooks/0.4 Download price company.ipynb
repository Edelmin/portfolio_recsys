##### Extracción de los ticker de las empresas

```python
# Funciones

def extract_tikr(file_name='file_name.xlsx'):
    
    return file_name.split("-")[1].strip()
    
def extract_currency(file_name='file_name.xlsx'):
    
    return file_name.split("-")[0].strip()
    
```

```python
from pathlib import Path


historico_datos_financieros = Path("C:/TFM/tfm_env/data/01_raw/01.3 - HistoricoEstadosFinancieros")

tickers_name_and_currencies_path = Path("C:/TFM/tfm_env/data/01_raw/01.3.1 tickers_and_currencies_per_sector/")


# lista_sectores = os.listdir(historico_datos_financieros.as_posix())

for sector_dir in os.listdir(historico_datos_financieros.as_posix()):
    for file in os.listdir(historico_datos_financieros.joinpath(sector_dir).as_posix()):

        info_file={
            "ticker": extract_tikr(file),
            "currency": extract_currency(file)
        }

        with open(tickers_name_and_currencies_path.as_posix() + f"/ticker_TIKR_and_currency_{sector_dir}.txt", mode="a") as file:
            
            file.write("{ticker} {currency}\n".format_map(info_file))
```
