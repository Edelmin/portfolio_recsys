```python
import yfinance as yf

def save_csv_historical_currency_data(ticker, saving_path="data/01_raw/DatosHistoricosPrecios/ParDivisas/"):
    
    data = yf.download(ticker, period='max')
    data.to_csv(f"{saving_path}{ticker}.csv", index_label="Date")
    
```
